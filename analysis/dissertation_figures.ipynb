```python
import matplotlib.pyplot as plt
import numpy as np

# =========================
# Input data from your table
# =========================
stages = [
    "Boltz-2 + ChimeraX + Biopython",
    "1_protein_min",
    "2_protein_mem",
    "3_system_min_coarse",
    "4_system_min_fine"
]

# Convert all wall times to hours
times_hours = [
    12/60 + 16/3600,          # 12 min 16 s
    7/60,                     # 7 min 0 s
    3 + 16/60 + 8/3600,       # 3 h 16 min 8 s
    14 + 7/60 + 14/3600,      # 14 h 7 min 14 s
    1 + 55/60 + 13/3600       # 1 h 55 min 13 s
]

# =========================
# Plot
# =========================
plt.rcParams.update({
    "font.family": "Arial",
    "font.size": 11,
    "axes.linewidth": 1.0,
    "xtick.direction": "out",
    "ytick.direction": "out"
})

fig, ax = plt.subplots(figsize=(10, 5.2))

y = np.arange(len(stages))
colors = ["#9ecae1", "#9ecae1", "#6baed6", "#2171b5", "#4292c6"]

bars = ax.barh(y, times_hours, color=colors, edgecolor="black", linewidth=0.6)

# Labels on bars
for bar, t in zip(bars, times_hours):
    ax.text(
        bar.get_width() + 0.15,
        bar.get_y() + bar.get_height()/2,
        f"{t:.2f} h",
        va="center",
        ha="left",
        fontsize=10
    )

# Axis formatting
ax.set_yticks(y)
ax.set_yticklabels(stages)
ax.invert_yaxis()  # top-to-bottom order
ax.set_xlabel("Wall time (hours)")
ax.set_title("Runtime distribution across pipeline stages")

# Clean style
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(axis="x", linestyle="--", linewidth=0.6, alpha=0.5)

plt.tight_layout()

# Save figure
output_path = r"C:\Users\SamanthaTu\Downloads\fgfr1_20lipids\pipeline_runtime_barchart.png"
plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.show()

print("Saved to:", output_path)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# =========================
# Data
# =========================
stages = [
    "Start",
    "Protein min\nfinal",
    "Coarse min\nstart",
    "Coarse min\nfinal",
    "Fine min\nstart",
    "Fine min\nfinal"
]

energy = np.array([
    96849706239.3,
    -128527.6,
    -2774538.6,
    -5000900.8,
    -5003041.3,
    -5003057.7
])

x = np.arange(len(stages))

# =========================
# Formatter
# =========================
def sci_fmt(x, pos):
    return f"{x:.2e}"

# =========================
# Style
# =========================
plt.rcParams.update({
    "font.family": "Times New Roman",   # 可改 Arial
    "font.size": 11,
    "axes.linewidth": 1.0,
    "xtick.direction": "out",
    "ytick.direction": "out",
    "xtick.major.width": 0.8,
    "ytick.major.width": 0.8
})

# =========================
# Figure with broken axis
# =========================
fig, (ax_top, ax_bottom) = plt.subplots(
    2, 1,
    sharex=True,
    figsize=(9.2, 6.0),
    gridspec_kw={"height_ratios": [1.1, 2.2], "hspace": 0.04}
)

line_color = "#1f4e79"
marker_face = "#3b82c4"
start_color = "#c0392b"

# Plot data
ax_top.plot(
    x, energy,
    color=line_color, linewidth=1.8,
    marker="o", markersize=5.5,
    markerfacecolor=marker_face,
    markeredgecolor=line_color
)
ax_bottom.plot(
    x, energy,
    color=line_color, linewidth=1.8,
    marker="o", markersize=5.5,
    markerfacecolor=marker_face,
    markeredgecolor=line_color
)

# Highlight initial point
ax_top.scatter(
    x[0], energy[0],
    color=start_color, s=38, zorder=6,
    edgecolors="black", linewidths=0.4
)

# Axis limits
ax_top.set_ylim(9.0e10, 1.0e11)
ax_bottom.set_ylim(-5.4e6, 3.5e5)

# Manual y ticks to match your original style
top_ticks = [9.00e10, 9.25e10, 9.50e10, 9.75e10, 1.00e11]
bottom_ticks = [0, -1e6, -2e6, -3e6, -4e6, -5e6]

ax_top.set_yticks(top_ticks)
ax_bottom.set_yticks(bottom_ticks)

ax_top.yaxis.set_major_formatter(FuncFormatter(sci_fmt))
ax_bottom.yaxis.set_major_formatter(FuncFormatter(sci_fmt))

# Remove spines
ax_top.spines["bottom"].set_visible(False)
ax_top.spines["top"].set_visible(False)
ax_top.spines["right"].set_visible(False)

ax_bottom.spines["top"].set_visible(False)
ax_bottom.spines["right"].set_visible(False)

# Tick settings
ax_top.tick_params(axis="x", which="both", bottom=False, labelbottom=False)
ax_bottom.tick_params(axis="x", rotation=35)

# Grid
ax_top.grid(axis="y", linestyle="--", linewidth=0.6, alpha=0.35)
ax_bottom.grid(axis="y", linestyle="--", linewidth=0.6, alpha=0.35)

# Labels
ax_bottom.set_xticks(x)
ax_bottom.set_xticklabels(stages, ha="right")
ax_top.set_ylabel("Energy (kJ/mol)")
ax_bottom.set_ylabel("Energy (kJ/mol)")
ax_bottom.set_xlabel("Stage")

# Broken-axis marks
d = 0.012
kwargs = dict(transform=ax_top.transAxes, color="black", clip_on=False, linewidth=1.0)
ax_top.plot((-d, +d), (-d, +d), **kwargs)
ax_top.plot((1 - d, 1 + d), (-d, +d), **kwargs)

kwargs.update(transform=ax_bottom.transAxes)
ax_bottom.plot((-d, +d), (1 - d, 1 + d), **kwargs)
ax_bottom.plot((1 - d, 1 + d), (1 - d, 1 + d), **kwargs)

# Optional annotation
ax_top.annotate(
    "Initial structure",
    xy=(x[0], energy[0]),
    xytext=(0.55, 9.78e10),
    textcoords="data",
    fontsize=10,
    arrowprops=dict(arrowstyle="-", lw=0.8, color="black"),
    ha="left"
)

# Margins
ax_bottom.margins(x=0.05)

plt.tight_layout()

# Save
output_path = r"C:\Users\SamanthaTu\Downloads\fgfr1_20lipids\energy_broken_axis_plot_readable.png"
plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.show()

print("Saved to:", output_path)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

npz_path = Path(r"C:\Users\SamanthaTu\Downloads\fgfr2_20lipids\pae_fgfr2_20lipids_model_0.npz")
protein_len = 821
out_png = npz_path.with_name("fgfr2_pae_protein_only_821aa_fixed.png")

data = np.load(npz_path)

pae = None
for key in ["pae", "predicted_aligned_error", "alignment_error", "predicted_alignment_error"]:
    if key in data.files:
        pae = data[key]
        print(f"Using key: {key}")
        break

if pae is None:
    raise ValueError(f"Could not find PAE matrix. Keys: {data.files}")

pae_protein = pae[:protein_len, :protein_len]

plt.figure(figsize=(8, 7), dpi=300)
im = plt.imshow(
    pae_protein,
    cmap="viridis",
    origin="upper",   # 關鍵：和原圖一致
    vmin=0,
    vmax=32,
    aspect="equal"
)

cbar = plt.colorbar(im)
cbar.set_label("PAE", fontsize=12)

plt.title("FGFR2 PAE heatmap", fontsize=16)
plt.xlabel("Residue index", fontsize=12)
plt.ylabel("Residue index", fontsize=12)

plt.tight_layout()
plt.savefig(out_png, bbox_inches="tight")
plt.show()

print(f"Saved to: {out_png}")
```
